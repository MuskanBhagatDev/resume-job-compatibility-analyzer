### Step 5: Domain Category Distribution Analysis
Comparison of employer demand (job categories) versus applicant supply (candidate inferred categories).

```python
job_categories = data_loader.get_job_category_distribution(job_df)
print("=== JOB POSTING CATEGORIES ===")
display(job_categories)

cv_categories = data_loader.get_candidate_category_distribution(cv_df)
print("\n=== CANDIDATE INFERRED CATEGORIES ===")
display(cv_categories)

# Render Category Plots
eda_plots.plot_job_category_distribution(job_df)
eda_plots.plot_candidate_category_distribution(cv_df)
eda_plots.plot_category_comparison(job_df, cv_df)
plt.show()
```

### Step 6: Compatibility Label Distribution (`label_strength`)

---
### Step 7: Target Variable Analysis — Continuous `final_score`
Regression target variable distribution, central tendencies (mean, median), spread (std, IQR), and skewness.

```python
print("Final Score Summary Statistics:")
final_stats = matches_df["final_score"].describe().to_frame().T
final_stats["skew"] = matches_df["final_score"].skew()
final_stats["kurtosis"] = matches_df["final_score"].kurt()
display(final_stats.round(4))

eda_plots.plot_final_score_distribution(matches_df)
plt.show()
```
